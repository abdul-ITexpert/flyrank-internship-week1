## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

The starter dataset contains 30,000 content records. Of these, 16,262 pages (about 54.2%) have a `down` trend direction. At the same time, page visibility varies considerably: the median number of 90-day impressions is 731, compared with a mean of about 5,200, with a maximum of 517,715 impressions. These numbers make content opportunity scoring worth investigating because a large number of pages show downward movement, while the importance of a page can vary substantially based on its level of search visibility. The goal is therefore to investigate whether observable signals can help prioritize pages for human review rather than treating every declining page equally.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)


```

```python
total_pages = len(df)
down_pages = (df["trend_direction"] == "down").sum()
down_pct = down_pages / total_pages * 100

median_impressions = df["impressions_90d"].median()
mean_impressions = df["impressions_90d"].mean()
max_impressions = df["impressions_90d"].max()

print(f"Total content records: {total_pages:,}")
print(f"Pages with downward trend: {down_pages:,} ({down_pct:.1f}%)")
print(f"Median 90-day impressions: {median_impressions:,.0f}")
print(f"Mean 90-day impressions: {mean_impressions:,.0f}")
print(f"Maximum 90-day impressions: {max_impressions:,.0f}")
```
